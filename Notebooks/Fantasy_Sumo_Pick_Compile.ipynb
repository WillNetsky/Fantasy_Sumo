```python
import pandas as pd
```

```python
month = '3'
year_short = '22'
```

```python
file1 = open(month+'-'+year_short+'-Picks.txt', 'r')
picks = []
count=0
while True:
    count += 1
    line = file1.readline()
    if len(line) == 1:
        continue
    if not line:
        break
    picks.append(line.strip())
```

```python
# Divide by 6 to get total entries
len(picks)/6
```

```python
dict = {}

for i in range(0,len(picks),6):
    poster = picks[i]
    y_o = picks[i+1].split(':')[1].strip()
    s_k = picks[i+2].split(':')[1].strip()
    m1_5 = picks[i+3].split(':')[1].strip()
    m6_10 = picks[i+4].split(':')[1].strip()
    m_11p = picks[i+5].split(':')[1].strip()
    
    dict[poster] = [y_o,s_k,m1_5,m6_10,m_11p]
```

```python
picks = pd.DataFrame.from_dict(dict,orient='index').reset_index()
picks.columns = ['Poster','YO','SK','M1-5','M6-10','M11+']
```

```python
goonzuke = pd.read_excel('Goonzuke-'+month+'-22.xlsx')
goonzuke.columns = ['Poster','gRank','gTier']
```

```python
dash = pd.merge(goonzuke,picks,on="Poster",how='right')
```

```python
new_juryo_rank = dash[dash.gTier == 'Juryo'].gRank.max()+1
```

```python
dash.gTier = dash.gTier.fillna('Juryo')
```

```python
dash.gRank = dash.gRank.fillna(new_juryo_rank)
```

```python
dash.to_csv(month+'-'+year_short+'-sumo-picks.csv',index=False)
```
